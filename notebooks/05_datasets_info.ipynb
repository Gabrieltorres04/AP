# Análise dos Datasets Externos
OTB (Meta + Google), Claude Multiround (Anthropic), RAID (OpenAI + Human), dataset_human_filtrado, HC3 Human

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = "../data/"
```

## Carregar datasets

```python
df_otb          = pd.read_csv(DATA_PATH + "dataset_OTB.csv")
df_claude       = pd.read_csv(DATA_PATH + "dataset_claude.csv")
df_raid         = pd.read_csv(DATA_PATH + "dataset_RAID.csv")
df_human_filt   = pd.read_csv(DATA_PATH + "dataset_human_filtrado.csv")
df_hc3          = pd.read_csv(DATA_PATH + "dataset_HC3_human.csv")

# Normalize column names to lowercase
for df in [df_otb, df_claude, df_raid, df_hc3]:
    df.rename(columns={"Text": "text", "Label": "label"}, inplace=True)

# dataset_human_filtrado already uses lowercase but column may be named differently
df_human_filt.rename(columns={"Text": "text", "Label": "label"}, inplace=True)

datasets = [
    ("OTB",            df_otb),
    ("Claude",         df_claude),
    ("RAID",           df_raid),
    ("Human filtrado", df_human_filt),
    ("HC3 Human",      df_hc3),
]

for name, df in datasets:
    print(f"{name:20s}: {df.shape}")
```

## Contagem por label

```python
for name, df in datasets:
    print(f"\n--- {name} ---")
    print(df["label"].value_counts())
```
